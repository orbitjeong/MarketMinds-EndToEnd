```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
      .str.replace("-", "_")
)

df.columns
```

```python
[c for c in df.columns if "spend" in c or "amt" in c or "sales" in c or "mnt" in c]
```

```python
spend_cols = ['mntwines',
 'mntfruits',
 'mntmeatproducts',
 'mntfishproducts',
 'mntsweetproducts',
 'mntbakedprods']
df["total_spend"] = df[spend_cols].sum(axis=1)
```

```python
df[['total_spend'] + spend_cols].head()
```
