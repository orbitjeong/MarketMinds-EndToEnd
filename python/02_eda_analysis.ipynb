## 2. Does campaign participation influence total spending?


```python
campaign_cols = [
    'acceptedcmp1',
    'acceptedcmp2',
    'acceptedcmp3',
    'acceptedcmp4',
    'acceptedcmp5',
    'response'   
]
```

```python
campaign_spend = {
    c: df.loc[df[c] == 1, 'total_spend'].mean()
    for c in campaign_cols
}

campaign_spend
```

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
pd.Series(campaign_spend).sort_values().plot(kind="barh")

plt.title("Average Total Spend by Campaign Participation")
plt.xlabel("Average Spend")
plt.ylabel("Campaign")

plt.show()
```
